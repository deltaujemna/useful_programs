## Komórka kodu zapisująca wynik sumy kwadratów (wraz ze zapisywaniem pliku tekstowego)

## Komórka kodu tworząca wykres (wraz ze wygenerowaniem obrazka)

````python
import matplotlib.pyplot as plt

# Wartości k od 1 do 20 i odpowiadające im wartości y = k^2
k = list(range(1, 21))

# Obliczanie wartości y = k^2 dla każdego k
y = [i**2 for i in k]

# Tworzenie wykresu
plt.figure(figsize=(8, 5))
plt.plot(k, y, marker="o", color="red", label=r"$y=k^2$")

plt.title("Wykres funkcji $y = k^2$")
plt.xlabel("k")
plt.ylabel("y")
plt.legend()
plt.tight_layout()
plt.grid()

# Zapis wykresu do pliku PNG oraz wyświetlenie go
plt.savefig("wykres.png", dpi=150, bbox_inches="tight")
plt.show()
````

## Komórka kodu wygenerująca animację wahadła (wraz ze zapisaniem video w formie .MP4 oraz .gif)

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Parametry wahadła
L = 1.0          # długość wahadła [m]
g = 9.81         # przyspieszenie ziemskie [m/s^2]
theta0 = np.pi/4 # wychylenie początkowe (45°)

# Czas
t = np.linspace(0, 10, 300)

# Przybliżone rozwiązanie dla małych wychyleń
omega = np.sqrt(g / L)
theta = theta0 * np.cos(omega * t)

# Położenie ciężarka
x = L * np.sin(theta)
y = -L * np.cos(theta)

# Przygotowanie wykresu
fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-1.2 * L, 1.2 * L)
ax.set_ylim(-1.2 * L, 0.2 * L)
ax.set_aspect('equal')
ax.grid(True)
ax.set_title("Animacja wahadła")

# Elementy animacji
line, = ax.plot([], [], 'b-', lw=2)
bob, = ax.plot([], [], 'ro', markersize=12)

def update(frame):
    line.set_data([0, x[frame]], [0, y[frame]])
    bob.set_data([x[frame]], [y[frame]])
    return line, bob

# Tworzenie animacji
anim = FuncAnimation(
    fig,
    update,
    frames=len(t),
    interval=30,
    blit=True
)

# Zapis do MP4
anim.save("video.mp4", writer="ffmpeg", fps=30)

print("Zapisano plik: video.mp4")
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Parametry wahadła
L = 1.0
g = 9.81
theta0 = np.pi / 4

# Czas
t = np.linspace(0, 10, 300)

# Ruch wahadła (przybliżenie dla małych wychyleń)
omega = np.sqrt(g / L)
theta = theta0 * np.cos(omega * t)

# Położenie ciężarka
x = L * np.sin(theta)
y = -L * np.cos(theta)

# Rysunek
fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-1.2 * L, 1.2 * L)
ax.set_ylim(-1.2 * L, 0.2 * L)
ax.set_aspect('equal')
ax.grid(True)
ax.set_title("Animacja wahadła")

line, = ax.plot([], [], 'b-', lw=2)
bob, = ax.plot([], [], 'ro', markersize=12)

def update(frame):
    line.set_data([0, x[frame]], [0, y[frame]])
    bob.set_data([x[frame]], [y[frame]])
    return line, bob

# Utworzenie animacji
anim = FuncAnimation(
    fig,
    update,
    frames=len(t),
    interval=30,
    blit=True
)

# Zapis do GIF
anim.save("video.gif", writer="pillow", fps=30)

print("Zapisano plik: video.gif")
````

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parametr
t = np.linspace(0, 10 * np.pi, 1000)

# Nowe równania parametryczne
x = (1 + 0.05 * t) * np.cos(t)
y = (1 + 0.05 * t) * np.sin(t)
z = 0.1 * t

# Wykres 3D
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, color='crimson', linewidth=2,
        label="Expanding Helix")

ax.set_title("3D Expanding Helix", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()

# Zapis wykresu do pliku PNG
plt.savefig("parametric_curve.png", dpi=300, bbox_inches="tight")
plt.show()
````
